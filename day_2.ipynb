# Strings

```python
full_name = "Habiba Abubakar"
message = "Hello " + full_name + "," + " " + "would you like to learn some Python today? "
print(message)
```

```python
name = "Muhammad aminu abubaKAR"
print(name.title())
print(name.upper())
print(name.lower())
```

```python
quote = "Ben Carson said 'Reading is the only way out of poverty'"
print(quote)
```

```python
author_name = "Ben Carson"
message_2 = author_name + " said 'Reading is the only way out of poverty'"
print(message_2)
```

```python
person_name = "\tMuhammad Aminu Abubakar\t"
person_name_1 = "Muhammad \nAminu \nAbubakar"
print(person_name)
print(person_name_1)
```

```python
print(person_name.rstrip())
print(person_name.lstrip())
print(person_name.strip())
```
